```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


def forwardPass(w1, w2, b, df):
    li = []

    for i in range(len(df)):
        y_pred = w1 * df.iat[i, 0] + w2 * df.iat[i, 1] + b
        li.append(y_pred)

    return li


# Subgradient of |w|
def sign_w(w):
    if w > 0:
        return 1
    elif w < 0:
        return -1
    else:
        return 0


# Gradient for w1
def find_gradient_w(df, arr, lam, old_w):
    sum_error = 0

    for i in range(len(df)):
        sum_error += df.iat[i, 0] * (df.iat[i, 2] - arr[i])

    return (-1 / len(df) * sum_error) + lam * sign_w(old_w)


# Gradient for w2
def find_gradient_w2(df, arr, lam, old_w):
    sum_error = 0

    for i in range(len(df)):
        sum_error += df.iat[i, 1] * (df.iat[i, 2] - arr[i])

    return (-1 / len(df) * sum_error) + lam * sign_w(old_w)


# Gradient for bias
def find_gradient_b(df, arr):
    sum_error = 0

    for i in range(len(df)):
        sum_error += df.iat[i, 2] - arr[i]

    return (-1 / len(df)) * sum_error


# Update parameters
def update_param(old_w1, old_w2, old_b,
                 grad_w1, grad_w2, grad_b, alpha):

    new_w1 = old_w1 - alpha * grad_w1
    new_w2 = old_w2 - alpha * grad_w2
    new_b = old_b - alpha * grad_b

    return new_w1, new_w2, new_b


# Dataset
df = pd.DataFrame({
    'X1': [-1, -1, 1, 1],
    'X2': [-1, 1, -1, 1],
    'Y': [-2.2, -1.8, 1.8, 2.2]
})


# Initial values
w1 = 0
w2 = 0.03333
b = 0

lam = 0.5
alpha = 0.1


# Store values
cost = []

w1_arr = [w1]
w2_arr = [w2]
b_arr = [b]

pred = []


# Training
for i in range(15):

    # Forward pass
    arr = np.array(forwardPass(w1, w2, b, df))

    # Store predictions
    pred.append(arr)

    # Calculate cost BEFORE updating weights
    mse_cost = (1 / (2 * len(df))) * np.sum(
        (df['Y'].values - arr) ** 2
    )

    lasso_cost = lam * (abs(w1) + abs(w2))

    total_cost = mse_cost + lasso_cost

    # Store cost
    cost.append(total_cost)


    # Calculate gradients
    grad_w1 = find_gradient_w(df, arr, lam, w1)
    grad_w2 = find_gradient_w2(df, arr, lam, w2)
    grad_b = find_gradient_b(df, arr)


    # Update parameters
    w1, w2, b = update_param(
        w1, w2, b,
        grad_w1, grad_w2, grad_b,
        alpha
    )


    # Store updated parameters
    w1_arr.append(w1)
    w2_arr.append(w2)
    b_arr.append(b)


    # Print
    print(
        f"Iteration {i + 1}: "
        f"w1 = {w1:.4f}, "
        f"w2 = {w2:.4f}, "
        f"b = {b:.4f}, "
        f"cost = {total_cost:.4f}"
    )


    # Stop if both weights become zero
    if w1 == 0 and w2 == 0:
        break


# =====================================================
# 1. TRAINING DATA VS REGRESSION PREDICTION
# =====================================================

plt.scatter(
    df['X1'],
    df['Y'],
    color='red',
    label='Training Data'
)

plt.plot(
    df['X1'],
    pred[-1],
    label='Regression'
)

plt.xlabel("X1")
plt.ylabel("Y")
plt.title("Training Data vs Regression")
plt.legend()

plt.show()


# =====================================================
# 2. COST VS ITERATIONS
# =====================================================

plt.scatter(
    range(1, len(cost) + 1),
    cost
)

plt.xlabel("Iteration")
plt.ylabel("Cost")
plt.title("Cost over Iterations")

plt.show()
```
