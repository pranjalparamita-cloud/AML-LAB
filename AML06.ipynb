```python
def chiSquare (li,chi_tested):
  row =[]
  col =[]
  row.append(sum(li[0]))
  row.append(sum(li[1]))
  col.append(sum([row[0] for row in li]))
  col.append(sum([row[1] for row in li]))
  e = []
  for i in range(2):
    for j in range(2):
      e.append(col[i] * row[j]/(sum(row)))


  o=[]
  for i in range(len(li)):
    for j in range(len(li[0])):
      o.append(li[i][j])

  chi = 0
  for i in range(len(o)):
    chi += ((o[i]-e[i])**2/e[i])

  if (chi<=chi_tested):
    print("Fail to reject null hypothesis")
  else:
    print("Rejected null hypoothesis")

```
