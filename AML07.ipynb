```python
import numpy as np

a = np.array([82,84,81,83,85])
b = np.array([85,87,86,84,88])
c = np.array([88,89,91,90,92])


mean_a = np.mean(a)
mean_b = np.mean(b)
mean_c = np.mean(c)


grand_mean = (mean_a+mean_b+mean_c)/3
s = ((mean_a-grand_mean)**2)*5+((mean_b-grand_mean)**2)*5+((mean_c-grand_mean)**2)*5
mean_s = s/2
SSW = np.sum((a-mean_a)**2)+np.sum((b-mean_b)**2)+np.sum((c-mean_c)**2)
mean_SSW = SSW/12
F = mean_s/mean_SSW


F_critical = 3.89
print("F_Stastical  : ",F)
if F < F_critical:
    print("Null hypothesis is accepted")
else:
    print("Null hypothesis is rejected")
```

```python
import numpy as np

matrix = np.array([[18, 7],
                   [14, 11],
                   [20, 5]])

F_critical = 5.991
m = 3
n = 2

tempMatrix = []
tempChi = []
row_totals = matrix.sum(axis=1)
col_totals = matrix.sum(axis=0)
grand_total = matrix.sum()

for i in range(m):
    temp1 = []
    for j in range(n):
        expected = (row_totals[i] * col_totals[j]) / grand_total
        temp1.append(expected)
        chi_val = ((matrix[i, j] - expected) ** 2) / expected
        tempChi.append(chi_val)
    tempMatrix.append(temp1)
calculated_chi = sum(tempChi)
print(f"\nThe calculated chi square value is: {calculated_chi:.4f}")

if calculated_chi < F_critical:
    print("The null hypothesis is accepted")
else:
    print("The null hypothesis is rejected")
```

```python
import numpy as np
x = np.array([2,3,4,5,6,7,8,9])
y = np.array([45,50,54,60,65,70,74,80])

x_mean = np.mean(x)
y_mean = np.mean(y)
n = len(x)

X = (x-x_mean)**2
Y = (y-y_mean)**2

r = ((np.sum(x*y)-n*x_mean*y_mean))/(np.sqrt(np.sum(X)*np.sum(Y)))
print("The co-relation is: ",r)
```
