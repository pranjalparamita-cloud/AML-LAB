```python
import numpy as np
import pandas as pd
```

```python
def forwardPass(w,b,df):
  li=[]
  for i in range(len(df)):
    li.append(w*df.iat[i,0]+b)
  return li

```

```python
def computeCost(df,arr):
  sum=0
  for i in range(len(df)):
    sum+= (df.iat[i,1] - arr[i])**2
  return sum/len(df)
```

```python
def find_gradient_w(df,arr):
  sum = 0
  for i in range(len(df)):
    sum+= (df.iat[i,0]*(df.iat[i,1]-arr[i]))
  return (-2)/len(df) * sum

```

```python
def find_gradient_b(df,arr):
  sum = 0
  for i in range(len(df)):
    sum+= ((df.iat[i,1]-arr[i]))
  return (-2)/len(df) * sum

```

```python
def update_param(old_w,old_b,grad_w,grad_b,alpha):
  new_w  =old_w -alpha*grad_w
  new_b = old_b -alpha*grad_b
  return new_w,new_b

```

```python
#step1
w = 0
b = 0
alpha = 0.1
```

```python
#step2
arr = np.array(forwardPass(w,b,df))
print(arr)
```

```python
#step3
cost_1 = computeCost(df,arr)
print(cost_1)
```

```python
#step4
grad_w = find_gradient_w(df,arr)
grad_b = find_gradient_b(df,arr)
print(grad_w,grad_b)
```

```python
w,b = update_param(w,b,grad_w,grad_b,alpha)
print(w,b)
```

```python
df = pd.DataFrame({'X':[1,2,3],
                    'Y':[2,3,4]})
w = 0
b = 0
alpha = 0.1
count = 0
cost =[]
pred = []
for i in range(100000):
  arr = np.array(forwardPass(w,b,df))
  pred.append(arr)
  cost.append(computeCost(df,arr))
  if(cost[-1] == 0):
    break
  grad_w = find_gradient_w(df,arr)
  grad_b = find_gradient_b(df,arr)
  w,b = update_param(w,b,grad_w,grad_b,alpha)
```
